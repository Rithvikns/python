```python
def rob_linear(arr):
  prev = curr = 0
  print(arr)
  for num in arr:
    print(prev,curr,num)
    prev , curr = curr , max(curr , prev+ num)
    print(prev,curr)
  return curr
```

```python
def rob(arr):
  if len(arr) == 0:
    return 0
  return max(rob_linear(arr[1:]), rob_linear(arr[:-1]))
```

```python
print(rob([3,2,7,10]))
```
